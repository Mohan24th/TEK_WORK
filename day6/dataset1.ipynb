```python
import pandas as pd
```

```python
import numpy as np
```

```python
df=pd.read_csv('Dataset_1.csv')
```

1 Identify all rows where order_amount is non-numeric but is_delivered is marked as delivered

```python
df['order_amount'].unique()
```

```python
mask=pd.to_numeric(df['order_amount'],errors='coerce').isna() 
df['is_delivered'].astype(str).str.lower().isin(["true","yes"])
df[mask]
```

2 Convert order_date to datetime and find owners where delivery_days is negative after valid date conversion

```python
df['order_date'].unique()
```

```python
df['order_date']=pd.to_datetime(df['order_date'],errors='coerce',dayfirst=True)
df['order_date']
```

```python
df['delivery_days']=pd.to_numeric(df['delivery_days'],errors='coerce')
```

```python
mask = (df['delivery_days'] < 0) & (df['order_date'].notna())
df[mask]
```

3 Standardize pyment_mode and calculate the percentage share of each mode per city

```python
df['payment_mode']=df['payment_mode'].str.strip().str.lower()
df['city']=df['city'].str.strip().str.lower()
```

```python
share= (df.groupby(['city', 'payment_mode']).size())
percentage=share/share.groupby(level=0).sum()*100
percentage_df=percentage.reset_index(name='percentage')
print(percentage_df)
```
